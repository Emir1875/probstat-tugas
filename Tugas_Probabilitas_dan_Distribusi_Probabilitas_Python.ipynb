# **1. Cathedral**

```python
!pip install cathedral
```

```python
from cathedral import model, flip, infer

@model
def animal():
    is_bird = flip(0.3, name="is_bird")
    if is_bird:
        can_fly = flip(0.8, name="can_fly")
        return {"type": "bird", "flies": can_fly}
    else:
        return {"type": "bukan_burung"}

# Inferensi posterior
posterior = infer(animal, method="importance", num_samples=1000)
print(posterior)
```

# **6. PyMC**

```python
import pymc as pm
import numpy as np
import arviz as az
import warnings
warnings.filterwarnings('ignore')

np.random.seed(42)
x_data = np.array([1, 2, 3, 4, 5])
y_data = 2.0 + 1.5 * x_data + np.random.normal(0, 0.1, size=5)

print("Memulai proses sampling model PyMC...")

with pm.Model() as model:
    alpha = pm.Normal('alpha', mu=0, sigma=10)
    beta = pm.Normal('beta', mu=0, sigma=10)
    sigma = pm.HalfNormal('sigma', sigma=1)

    mu = alpha + beta * x_data
    y_obs = pm.Normal('y_obs', mu=mu, sigma=sigma, observed=y_data)

    trace = pm.sample(
        draws=200,
        tune=200,
        chains=1,
        cores=1,
        progressbar=False,
        random_seed=42
    )

print("\n=== RINGKASAN HASIL MODEL PYMC ===\n")
summary = az.summary(trace, var_names=["alpha", "beta", "sigma"])
print(summary)

print("\nCatatan:")
print("Nilai 'mean' pada baris 'alpha' seharusnya mendekati 2.0.")
print("Nilai 'mean' pada baris 'beta' seharusnya mendekati 1.5.")
```
